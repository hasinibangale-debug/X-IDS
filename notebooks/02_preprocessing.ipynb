```python
# Search recursively inside all subfolders
flow_csv_files = sorted(
    glob.glob(os.path.join(FLOW_DIR, "**", "*.csv"), recursive=True)
)

ml_csv_files = sorted(
    glob.glob(os.path.join(ML_DIR, "**", "*.csv"), recursive=True)
)

print("GeneratedLabelledFlows files:", len(flow_csv_files))
print("MachineLearningCSV files:", len(ml_csv_files))

print("\nGeneratedLabelledFlows:")
for file in flow_csv_files:
    print(os.path.basename(file))

print("\nMachineLearningCSV:")
for file in ml_csv_files:
    print(os.path.basename(file))
```

```python
# Dictionary to store all CSV files
dataframes = {}

for file in flow_csv_files:

    filename = os.path.basename(file)

    print("Loading:", filename)

    df = pd.read_csv(
        file,
        encoding="cp1252",
        low_memory=False
    )

    # Remove whitespace from column names
    df.columns = df.columns.str.strip()

    # Remove completely empty rows
    df = df.dropna(how="all").reset_index(drop=True)

    dataframes[filename] = df

    print("Shape:", df.shape)

print("\nAll files loaded successfully!")

print("\nTotal records:",
      sum(len(df) for df in dataframes.values()))
```

```python
import os
import pandas as pd

RAW_CHECKPOINT_DIR = os.path.join(
    CHECKPOINT_DIR, "source_parquet"
)

os.makedirs(RAW_CHECKPOINT_DIR, exist_ok=True)

for filename in sorted(os.listdir(FLOW_DIR)):

    if not filename.endswith(".csv"):
        continue

    csv_path = os.path.join(FLOW_DIR, filename)

    parquet_name = filename.replace(".csv", ".parquet")
    parquet_path = os.path.join(
        RAW_CHECKPOINT_DIR, parquet_name
    )

    print("\nProcessing:", filename)

    chunks = []

    for chunk in pd.read_csv(
        csv_path,
        encoding="cp1252",
        low_memory=False,
        chunksize=50000
    ):
        chunk.columns = chunk.columns.str.strip()

        # Remove fully blank rows
        chunk = chunk.dropna(how="all")

        if not chunk.empty:
            chunks.append(chunk)

    file_df = pd.concat(chunks, ignore_index=True)

    file_df.to_parquet(
        parquet_path,
        index=False
    )

    print("Saved rows:", len(file_df))
    print("Saved to:", parquet_name)

    del file_df, chunks
```

```python

import os
import numpy as np
from datetime import datetime

# Load saved labels
y_train = np.load(os.path.join(PROCESSED_DIR, "y_train.npy"))
y_val = np.load(os.path.join(PROCESSED_DIR, "y_val.npy"))
y_test = np.load(os.path.join(PROCESSED_DIR, "y_test.npy"))

# Calculate class distribution
all_labels = sorted(np.unique(y_train).tolist())

class_rows = []
for label in all_labels:
    train_count = int(np.sum(y_train == label))
    val_count = int(np.sum(y_val == label))
    test_count = int(np.sum(y_test == label))

    class_rows.append(
        f"| {label} | {train_count + val_count + test_count:,} "
        f"| {train_count:,} | {val_count:,} | {test_count:,} |"
    )

report = f"""# X-IDS – Data Preprocessing Report

## 1. Dataset Overview

- **Dataset:** CIC-IDS2017
- **Selected subset:** GeneratedLabelledFlows
- **Total records after cleaning:** 2,830,540
- **Selected numerical features:** 64
- **Traffic classes:** 15

## 2. Data Cleaning

The following operations were performed:

1. Replaced positive and negative infinity values with missing values.
2. Removed 8 constant all-zero columns.
3. Removed 5 duplicate columns/features.
4. Converted the Timestamp column to datetime format.
5. Replaced invalid negative measurements with missing values in 10 measurement columns.
6. Preserved -1 sentinel values in the initial TCP window-size features.
7. Removed 203 exact duplicate rows.
8. Created a Capture_Date helper column.
9. Applied median imputation to missing numerical feature values.

## 3. Feature Selection

- **Selected features:** 64 numerical features.
- Metadata excluded from the initial Random Forest feature matrix:
  Flow ID, Source IP, Source Port, Destination IP,
  Destination Port, Protocol, Timestamp, Label and Capture_Date.
- The original metadata remains available in cleaned_flows.parquet.

## 4. Train, Validation and Test Split

A stratified split was performed using random_state = 42.

| Dataset | Records | Percentage |
|---|---:|---:|
| Training | {len(y_train):,} | 70% |
| Validation | {len(y_val):,} | 15% |
| Testing | {len(y_test):,} | 15% |
| **Total** | **{len(y_train) + len(y_val) + len(y_test):,}** | **100%** |

## 5. Class Distribution

| Traffic Class | Total | Training | Validation | Testing |
|---|---:|---:|---:|---:|
{chr(10).join(class_rows)}

## 6. Missing Value Handling

- Strategy: Median imputation.
- The imputer was fitted only on the training data.
- The same fitted imputer was applied to validation and test data.
- All three saved feature matrices were verified to contain zero missing values.

## 7. Generated Files

- X_train.npy
- X_val.npy
- X_test.npy
- y_train.npy
- y_val.npy
- y_test.npy
- cleaned_flows.parquet
- data_split_indices.npz
- feature_columns.joblib
- median_imputer.joblib
- label_mapping.json
- preprocessing_config.json

## 8. Important Notes and Limitations

- The dataset has a highly imbalanced class distribution.
- Some classes contain very few records, so their evaluation metrics may be unstable.
- The random stratified split does not guarantee separation of related network hosts or traffic sessions across the three subsets.
- Source and destination IP addresses and timestamps are retained in the cleaned dataset for subsequent graph construction.
- Validation data should be used for model selection. The test set should be reserved for final evaluation.

## 9. Preprocessing Status

Dataset cleaning, feature selection, stratified splitting,
training-only median imputation and output verification
have been completed.

The generated datasets and metadata are stored in the
X-IDS data/processed directory.
"""

report_path = os.path.join(
    PROCESSED_DIR, "preprocessing_report.md"
)

with open(report_path, "w", encoding="utf-8") as file:
    file.write(report)

print("Preprocessing report saved successfully!")
print("File location:", report_path)
print("\nReport size:", os.path.getsize(report_path), "bytes")
```
